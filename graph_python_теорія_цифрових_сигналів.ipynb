# *MATPLOTLIB*

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-2, 2, 25)
y= x**2

figure, ax = plt.subplots(figsize = (7,7))
plt.plot(x,y, label =  "nigerfirst", color = 'black', linestyle = '-.', linewidth = 2.5, markerfacecolor = 'yellow', marker = '*', markersize = '8')
ax.axhline(y = 0, color='r', linestyle='--')
ax.axvline(x = 0, color='r', linestyle='--')
ax.grid(color = 'b', linestyle = ':')
plt.xlim(-5,5)
plt.ylim(-2,5)

plt.title("2 graphs")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def f(t):
  return np.exp(-t) * np.cos(2*np.pi*t)

t1 = np.arange(0.0, 50, 0.1)
t2 = np.arange(0.0, 30, 0.02)

plt.figure(figsize=(12,7))
plt.subplot(211)

plt.plot(t1, f(t1), color='tab:blue', marker= 'o')
plt.plot(t2, f(t2), color='black')

plt.subplot(212)
plt.plot(t2, np.cos(2*np.pi*t2), color='orange', linestyle = '--')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

fs = 44100
T=1
t = np.linspace(0, T, fs * T, endpoint=False)

#samples
A1, A2, A3 = 10, 10, 10
F1, F2, F3 = 261.63, 329.63, 440

x1 = A1*np.sin(2*np.pi*t*F1)
x2 = A2*np.sin(2*np.pi*t*F2)
x3 = A3*np.sin(2*np.pi*t*F3)

X = x1+x2+x3

plt.figure(figsize=(8, 10))

#GRAPH PAINTING
plt.suptitle('C-major-chord')

plt.subplot(4,1,1)
plt.plot(t, x1, color='red')
plt.plot(t, x2, color='green')
plt.plot(t, x3, color='blue')
plt.grid()
plt.xlim(0,0.005)
plt.ylabel('Amplitude')

plt.subplot(4,1,2)
plt.plot(t, x1, color='red')
plt.plot(t, x2, color='green')
plt.plot(t, x3, color='blue')
plt.grid()
plt.xlim(0,0.05)
plt.ylabel('Amplitude')

plt.subplot(4,1,3)
plt.plot(t, X)
plt.grid()
plt.xlim(0,0.05)
plt.ylabel('Amplitude')

plt.subplot(4,1,4)
plt.plot(t, X, color = '#C82909')
plt.grid()
plt.xlim(0,1)
plt.xlabel('Time')
plt.ylabel('Amplitude')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

fs = 44100
T=1
t = np.linspace(0, T, fs * T, endpoint=False)

#samples
A1, A2, A3 = 15, 6, 5
F1, F2, F3 = 1, 15, 50

x1 = A1*np.sin(2*np.pi*t*F1)
x2 = A2*np.sin(2*np.pi*t*F2)
x3 = A3*np.sin(2*np.pi*t*F3)

X = x1+x2+x3

plt.figure(figsize=(8, 10))

plt.subplot(3,1,1)
plt.title('Amplitude = 15 V, Frequency = 1 Hz')
plt.plot(t, x1)
plt.xlabel('t(s)')
plt.ylabel('U(V)')
plt.grid()

plt.subplot(3,1,2)
plt.title('Amplitude = 6 V, Frequency = 15 Hz')
plt.plot(t, x2)
plt.xlabel('t(s)')
plt.ylabel('U(V)')
plt.grid()

plt.subplot(3,1,3)
plt.title('Amplitude = 5 V, Frequency = 50 Hz')
plt.plot(t, x3)
plt.xlabel('t(s)')
plt.ylabel('U(V)')
plt.grid()

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import random
from scipy import signal

fs = 256
T=1
t = np.linspace(0, T, fs * T, endpoint=False)

print("choose option\n1. enter custom amplitudes\n2. do random amplitudes")
choose = int(input("Enter choice (1 or 2): "))

match choose:
    case 2:
        A1, A2, A3 = [random.randrange(1, 10, 1) for i in range(3)]
    case 1:
        A1 = float(input("Enter A1: "))
        A2 = float(input("Enter A2: "))
        A3 = float(input("Enter A3: "))
    case _:
        print("invalid choose. A1, A2, A3 = 15, 6, 5 ")
        A1, A2, A3 = 15, 6, 5

print("\nDo you want to enter custom frequencies?")
print("1. Yes (enter manually) \n2. No (use default 1, 15, 50 Hz)")
choose_f = int(input("Enter choice (1 or 2): "))

match choose_f:
    case 1:
        F1 = float(input("Enter Frequency F1 (Hz): "))
        F2 = float(input("Enter Frequency F2 (Hz): "))
        F3 = float(input("Enter Frequency F3 (Hz): "))
    case _:
        print("Using default frequencies: 1, 15, 50 Hz")
        F1, F2, F3 = 1, 15, 50

Ap = A1
Fp = 5
duty = 0.5
fc = 10
width = 0.2
centre = 0.5
PF = int(fs*(centre-width/2))
PS = int(fs*(centre+width/2))

x1 = A1*np.sin(2*np.pi*t*F1)
x2 = A2*np.sin(2*np.pi*t*F2)
x3 = A3*np.sin(2*np.pi*t*F3)
X = x1+x2+x3

rect_pulses = Ap*signal.square(2*np.pi*Fp*t, duty=duty)
tri_pulses = Ap*signal.sawtooth(2*np.pi*Fp*t, width=duty)
gauss_pulse = Ap*signal.gausspulse(t - T/2, fc=fc)

single_pulse = np.zeros(np.size(t))
single_pulse[PF:PS] = Ap

plt.figure(figsize=(8, 10))

plt.subplot(3,1,1)
plt.title(f'Amplitude = {A1} V, Frequency = {F1} Hz')
plt.plot(t, x1)
plt.xlabel('t(s)')
plt.ylabel('U(V)')
plt.grid()

plt.subplot(3,1,2)
plt.title(f'Amplitude = {A2} V, Frequency = {F2} Hz')
plt.plot(t, x2)
plt.xlabel('t(s)')
plt.ylabel('U(V)')
plt.grid()

plt.subplot(3,1,3)
plt.title(f'Amplitude = {A3} V, Frequency = {F3} Hz')
plt.plot(t, x3)
plt.xlabel('t(s)')
plt.ylabel('U(V)')
plt.grid()

plt.tight_layout()

plt.figure(figsize=(8, 12))

plt.subplot(4,1,1)
plt.title(f'Rectangular pulses: Amplitude = {Ap} V, Frequency = {Fp} Hz, duty = {duty}')
plt.plot(t, rect_pulses)
plt.xlabel('t(s)')
plt.ylabel('U(V)')
plt.grid()

plt.subplot(4,1,2)
plt.title(f'Triangular pulses: Amplitude = {Ap} V, Frequency = {Fp} Hz')
plt.plot(t, tri_pulses)
plt.xlabel('t(s)')
plt.ylabel('U(V)')
plt.grid()

plt.subplot(4,1,3)
plt.title(f'Gaussian pulse: Amplitude = {Ap} V, fc = {fc} Hz')
plt.plot(t, gauss_pulse)
plt.xlabel('t(s)')
plt.ylabel('U(V)')
plt.grid()

plt.subplot(4,1,4)
plt.title(f'single pulse: Amplitude = {Ap} V, width = {width} s')
plt.plot(t, single_pulse)
plt.xlabel('t(s)')
plt.ylabel('U(V)')
plt.grid()

plt.tight_layout()
plt.show()
```
