# *MATPLOTLIB*

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-2, 2, 25)
y= x**2

figure, ax = plt.subplots(figsize = (7,7))
plt.plot(x,y, label =  "nigerfirst", color = 'black', linestyle = '-.', linewidth = 2.5, markerfacecolor = 'yellow', marker = '*', markersize = '8')
ax.axhline(y = 0, color='r', linestyle='--')
ax.axvline(x = 0, color='r', linestyle='--')
ax.grid(color = 'b', linestyle = ':')
plt.xlim(-5,5)
plt.ylim(-2,5)

plt.title("2 graphs")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def f(t):
  return np.exp(-t) * np.cos(2*np.pi*t)

t1 = np.arange(0.0, 50, 0.1)
t2 = np.arange(0.0, 30, 0.02)

plt.figure(figsize=(12,7))
plt.subplot(211)

plt.plot(t1, f(t1), color='tab:blue', marker= 'o')
plt.plot(t2, f(t2), color='black')

plt.subplot(212)
plt.plot(t2, np.cos(2*np.pi*t2), color='orange', linestyle = '--')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

fs = 44100
T=1
t = np.linspace(0, T, fs * T, endpoint=False)

#samples
A1, A2, A3 = 10, 10, 10
F1, F2, F3 = 261.63, 329.63, 440

x1 = A1*np.sin(2*np.pi*t*F1)
x2 = A2*np.sin(2*np.pi*t*F2)
x3 = A3*np.sin(2*np.pi*t*F3)

X = x1+x2+x3

plt.figure(figsize=(8, 10))

#GRAPH PAINTING
plt.subplot(4,1,1)
plt.plot(t, x1, color='red')
plt.plot(t, x2, color='green')
plt.plot(t, x3, color='blue')
plt.grid()
plt.xlim(0,0.005)
plt.ylabel('Amplitude')

plt.subplot(4,1,2)
plt.plot(t, x1, color='red')
plt.plot(t, x2, color='green')
plt.plot(t, x3, color='blue')
plt.grid()
plt.xlim(0,0.05)
plt.ylabel('Amplitude')

plt.subplot(4,1,3)
plt.plot(t, X)
plt.grid()
plt.xlim(0,0.05)
plt.ylabel('Amplitude')

plt.subplot(4,1,4)
plt.plot(t, X, color = '#69457D')
plt.grid()
plt.xlim(0,1)
plt.xlabel('Time')
plt.ylabel('Amplitude')

plt.tight_layout()
plt.show()
```
